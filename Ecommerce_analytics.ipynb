```python
products["Unit Cost USD"] = (
    products["Unit Cost USD"]
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
    .astype(float)
)

products["Unit Price USD"] = (
    products["Unit Price USD"]
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
    .astype(float)
)
```

```python
sales_analysis = sales.merge(
    products,
    on="ProductKey",
    how="left"
)
```

```python
sales_analysis.shape
```

```python
sales_analysis.head()
```

```python
sales_analysis = sales_analysis.merge(
    customers,
    on="CustomerKey",
    how="left"
)
```

```python
sales_analysis.shape
```

```python
sales_analysis = sales_analysis.merge(
    stores,
    on="StoreKey",
    how="left"
)
```

```python
sales_analysis.shape
```

```python
sales_analysis.columns
```

```python
sales_analysis.head(3)
```

```python
sales_analysis = sales_analysis.rename(columns={
    "Country_x": "Customer Country",
    "Country_y": "Store Country",
    "State_x": "Customer State",
    "State_y": "Store State"
})
```

```python
sales_analysis.columns
```

```python
sales_analysis["Revenue USD"] = (
    sales_analysis["Quantity"] *
    sales_analysis["Unit Price USD"]
)
```

```python
sales_analysis[
    ["Quantity", "Unit Price USD", "Revenue USD"]
].head()
```

```python
sales_analysis["Cost USD"] = (
    sales_analysis["Quantity"] *
    sales_analysis["Unit Cost USD"]
)
```

```python
sales_analysis["Profit USD"] = (
    sales_analysis["Revenue USD"] -
    sales_analysis["Cost USD"]
)
```

```python
sales_analysis["Profit Margin %"] = (
    sales_analysis["Profit USD"] /
    sales_analysis["Revenue USD"]
) * 100
```

```python
sales_analysis[
    ["Revenue USD", "Cost USD", "Profit USD", "Profit Margin %"]
].head()
```

```python
sales_analysis["Profit Margin %"] = sales_analysis["Profit Margin %"].round(2)
```

```python
sales_analysis[
    ["Quantity", "Revenue USD", "Cost USD", "Profit USD", "Profit Margin %"]
].describe()
```

```python
category_summary = sales_analysis.groupby("Category").agg(
    Revenue=("Revenue USD", "sum"),
    Cost=("Cost USD", "sum"),
    Profit=("Profit USD", "sum")
)
```

```python
sales_analysis["Order Date"].dtype
```

```python
sales_analysis["Year"] = sales_analysis["Order Date"].dt.year
sales_analysis["Month"] = sales_analysis["Order Date"].dt.month
sales_analysis["Month Name"] = sales_analysis["Order Date"].dt.month_name()
sales_analysis["Quarter"] = sales_analysis["Order Date"].dt.quarter
```

```python
yearly_sales = sales_analysis.groupby("Year").agg(
    Revenue=("Revenue USD", "sum"),
    Cost=("Cost USD", "sum"),
    Profit=("Profit USD", "sum"),
    Quantity=("Quantity", "sum")
).reset_index()

yearly_sales["Profit Margin %"] = (
    yearly_sales["Profit"] /
    yearly_sales["Revenue"] * 100
).round(2)

yearly_sales
```

```python
monthly_sales = sales_analysis.groupby(
    ["Year", "Month"]
).agg(
    Revenue=("Revenue USD", "sum"),
    Profit=("Profit USD", "sum")
).reset_index()

monthly_sales
```

```python
customer_summary = sales_analysis.groupby(
    ["CustomerKey", "Name", "City", "Customer State", "Customer Country"]
).agg(
    Orders=("Order Number", "nunique"),
    Quantity=("Quantity", "sum"),
    Revenue=("Revenue USD", "sum"),
    Profit=("Profit USD", "sum")
).reset_index()
```

```python
customer_summary["Profit Margin %"] = (
    customer_summary["Profit"] /
    customer_summary["Revenue"] * 100
).round(2)
```

```python
customer_summary.sort_values(
    "Revenue",
    ascending=False
).head(10)
```

```python
customer_summary.sort_values(
    "Profit",
    ascending=False
).head(10)
```

```python
customer_summary["AOV"] = (
    customer_summary["Revenue"] /
    customer_summary["Orders"]
).round(2)
```

```python
customer_summary.sort_values(
    "AOV",
    ascending=False
).head(10)
```

```python
country_summary = sales_analysis.groupby(
    "Customer Country"
).agg(
    Revenue=("Revenue USD", "sum"),
    Profit=("Profit USD", "sum"),
    Quantity=("Quantity", "sum"),
    Customers=("CustomerKey", "nunique")
).reset_index()

country_summary["Profit Margin %"] = (
    country_summary["Profit"] /
    country_summary["Revenue"] * 100
).round(2)

country_summary.sort_values(
    "Revenue",
    ascending=False
)
```

```python
store_summary = sales_analysis.groupby(
    ["StoreKey", "Store Country", "Store State"]
).agg(
    Revenue=("Revenue USD", "sum"),
    Profit=("Profit USD", "sum"),
    Quantity=("Quantity", "sum"),
    Orders=("Order Number", "nunique")
).reset_index()

store_summary["Profit Margin %"] = (
    store_summary["Profit"] /
    store_summary["Revenue"] * 100
).round(2)
```

```python
store_summary.sort_values(
    "Revenue",
    ascending=False
).head(10)
```

```python
total_revenue = sales_analysis["Revenue USD"].sum()
total_cost = sales_analysis["Cost USD"].sum()
total_profit = sales_analysis["Profit USD"].sum()
total_quantity = sales_analysis["Quantity"].sum()
total_orders = sales_analysis["Order Number"].nunique()
total_customers = sales_analysis["CustomerKey"].nunique()

overall_margin = (
    total_profit / total_revenue * 100
)
```

```python
print("Total Revenue:", total_revenue)
print("Total Cost:", total_cost)
print("Total Profit:", total_profit)
print("Profit Margin:", round(overall_margin, 2), "%")
print("Units Sold:", total_quantity)
print("Orders:", total_orders)
print("Customers:", total_customers)
```
